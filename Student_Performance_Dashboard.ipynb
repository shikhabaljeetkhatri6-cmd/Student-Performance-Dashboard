```python
# Create DataFrame
import pandas as pd

data = {
    "Student_ID": [
        "STU-001", "STU-002", "STU-003", "STU-004", "STU-005",
        "STU-006", "STU-007", "STU-008", "STU-009", "STU-010",
        "STU-011", "STU-012", "STU-013", "STU-014", "STU-015"
    ],
    "Gender": [
        "Female", "Male", "Female", "Male", "Female",
        "Male", "Female", "Male", "Female", "Male",
        "Female", "Male", "Female", "Male", "Female"
    ],
    "Lunch": [
        "standard", "standard", "free", "standard", "standard",
        "free", "standard", "free", "standard", "standard",
        "free", "standard", "standard", "free", "standard"
    ],
    "Test_Prep": [
        "none", "completed", "none", "none", "completed",
        "none", "completed", "none", "completed", "completed",
        "none", "none", "completed", "none", "completed"
    ],
    "Math_Score": [
        72, 69, 47, 76, 71, 40, 88, 64,
        78, 92, 38, 58, 87, 52, 90
    ],
    "Reading_Score": [
        74, 90, 54, 78, 83, 43, 95, 60,
        91, 89, 48, 62, 93, 57, 96
    ],
    "Writing_Score": [
        74, 88, 47, 75, 78, 39, 92, 58,
        90, 88, 44, 60, 89, 55, 94
    ]
}

df = pd.DataFrame(data)

df["Average_Score"] = (
    (df["Math_Score"] + df["Reading_Score"] + df["Writing_Score"]) / 3
).round(1)

print(df)
```

```python
# 1. Display first five rows - head()
print(df.head())
```

```python
# 2. Display last five rows - tail()
print(df.tail())
```

```python
# 3. Check DataFrame shape - shape
print(df.shape)
```

```python
# 4. Display column names - columns
print(df.columns)
```

```python
# 5. Get dataset information - info()
df.info()
```

```python
# 6. Statistical summary - describe()
print(df[["Math_Score", "Reading_Score", "Writing_Score"]].describe())
```

```python
# 7. Select students who completed test prep - filtering
print(df[df["Test_Prep"] == "completed"][["Student_ID", "Math_Score", "Average_Score"]])
```

```python
# 8. Sort average score in descending order - sort_values()
print(df.sort_values("Average_Score", ascending=False)[["Student_ID", "Average_Score"]].head())
```

```python
# 9. Average score by gender - groupby()
print(df.groupby("Gender")["Average_Score"].mean().round(1))
```

```python
# 10. Create a new column - Total_Score
df["Total_Score"] = df["Math_Score"] + df["Reading_Score"] + df["Writing_Score"]
print(df[["Student_ID", "Total_Score"]].head())
```

```python
import matplotlib.pyplot as plt
```

```python
# 1. Bar Chart - Average Score by Test Preparation
avg_prep = df.groupby("Test_Prep")["Average_Score"].mean()
plt.bar(avg_prep.index, avg_prep.values)
plt.title("Average Score by Test Preparation")
plt.xlabel("Test Preparation")
plt.ylabel("Average Score")
plt.tight_layout()
plt.show()
```

```python
# 2. Pie Chart - Lunch Type Share
lunch_count = df["Lunch"].value_counts()
plt.pie(lunch_count.values, labels=lunch_count.index,
        autopct="%1.1f%%", startangle=90)
plt.title("Share of Students by Lunch Type")
plt.show()
```

```python
# 3. Line Chart - Average Score of Each Student
plt.plot(df["Student_ID"], df["Average_Score"], marker="o")
plt.title("Average Score of Each Student")
plt.xlabel("Student ID")
plt.ylabel("Average Score")
plt.xticks(rotation=60)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# 4. Histogram - Average Score Distribution
plt.hist(df["Average_Score"], bins=5, edgecolor="black")
plt.title("Distribution of Average Score")
plt.xlabel("Average Score")
plt.ylabel("Number of Students")
plt.show()
```

```python
# 5. Scatter Plot - Math vs Reading Score
plt.scatter(df["Math_Score"], df["Reading_Score"])
plt.title("Math vs Reading Score")
plt.xlabel("Math Score")
plt.ylabel("Reading Score")
plt.grid(True)
plt.show()
```

```python
# 6. Box Plot - Score Spread by Subject
plt.boxplot([df["Math_Score"], df["Reading_Score"], df["Writing_Score"]])
plt.xticks([1, 2, 3], ["Math", "Reading", "Writing"])
plt.title("Score Spread by Subject")
plt.xlabel("Subject")
plt.ylabel("Score")
plt.grid(True)
plt.show()
```
